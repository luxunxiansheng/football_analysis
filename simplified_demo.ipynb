# 🏈 Football Analysis Pipeline - Simplified Demo

This notebook demonstrates the new unified `FootballAnalysisPipeline` that orchestrates all processors automatically.

```python
# Import the new pipeline
from football_ai import FootballAnalysisPipeline, get_default_config
import os
```

## Configuration Setup

The new system uses a comprehensive configuration system with presets.

```python
# Create configuration
config = get_default_config()

# Update paths
config.update_paths(
    model_path="models/detect/best.pt",
    input_video_path="input_videos/08fd33_4.mp4",
    output_video_path="outputs/videos/simplified_demo.mp4"
)

# Verify files exist
assert os.path.exists(config.model.player_model_path), f"Model not found: {config.model.player_model_path}"
assert os.path.exists(config.processing.input_video_path), f"Video not found: {config.processing.input_video_path}"

print("✅ Configuration ready")
print(config.get_summary())
```

```python
# Enable progress bars for better user experience
config.show_progress_bars = True
print("📊 Progress bars enabled - you'll see detailed progress for each processing stage!")
```

## Pipeline Execution

The pipeline automatically runs all processors in the correct order.

```python
# Initialize pipeline
pipeline = FootballAnalysisPipeline(config=config)

print(f"🚀 Pipeline ready with {len(pipeline.processors)} processors:")
for i, processor in enumerate(pipeline.processors):
    print(f"  {i+1}. {processor.__class__.__name__}")
```

```python
# Process the video
print("🎬 Starting video analysis...")

results = pipeline.process_video(
    video_path=config.processing.input_video_path,
    output_path=config.processing.output_video_path
)

print("✅ Analysis complete!")
```

## Results Summary

View comprehensive analysis results.

```python
# Get analysis summary
summary = pipeline.get_analysis_summary(results)

print("📊 Analysis Summary")
print("=" * 50)

# Video info
video_info = summary['video_info']
print(f"📹 Video: {os.path.basename(video_info['path'])}")
print(f"   Duration: {video_info['duration']:.1f}s")
print(f"   Frames: {video_info['frames_processed']}")
print(f"   Resolution: {video_info['resolution']}")
print()

# Detection summary
detections = summary['detection_summary']
print("🎯 Detection Summary:")
for obj_type, count in detections.items():
    print(f"   {obj_type}: {count}")
print()

# Tracking summary
tracking = summary['tracking_summary']
print(f"🏃 Tracking: {tracking['unique_tracks']} unique objects tracked")
print()

# Team summary
teams = summary['team_summary']
print("👕 Team Assignment:")
print(f"   Team 1: {teams['team_1']} assignments")
print(f"   Team 2: {teams['team_2']} assignments")
print(f"   Unknown: {teams['unknown']} assignments")
```

## Save Results

Save complete analysis results for later use.

```python
# Save results
output_dir = "outputs/data"
os.makedirs(output_dir, exist_ok=True)

results_path = f"{output_dir}/pipeline_results.pkl"
pipeline.save_results(results, results_path)

print(f"💾 Results saved to: {results_path}")
print(f"🎬 Output video: {config.processing.output_video_path}")
print("\n🎉 Demo complete!")
```
